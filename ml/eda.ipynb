```python
import matplotlib.pyplot as plt
import pandas as pd

def display_count_plots(columns, num_row, num_col, url):
    data_frame = pd.read_csv(url)

    # Check if the 'Attrition' column exists
    if "Attrition" not in data_frame.columns:
        print("Error: 'Attrition' column not found in the DataFrame.")
        return

    # Create subplots (4 rows, 3 columns)
    _, axes = plt.subplots(num_row, num_col, figsize=(18, 20))

    for i in range(len(columns)):
        if columns[i] not in data_frame.columns:
            print(f"Warning: Column '{columns[i]}' not found in the DataFrame.")
            continue  

        row = i // num_col
        col = i % num_col

        churn = data_frame.groupby([columns[i], "Attrition"]).size().unstack()

        if churn.empty:
            axes[row, col].axis('off')
        else:
            churn.plot(kind="bar", stacked=False, ax=axes[row, col], color=["#d62728", "#17becf"], alpha=0.7)

            axes[row, col].set_title(f'Attrition by {columns[i]}', fontsize=14)
            axes[row, col].set_xlabel(columns[i], fontsize=12)
            axes[row, col].set_ylabel('Count', fontsize=12)
            axes[row, col].tick_params(axis='x', rotation=0, labelsize=10)

    for i in range(len(columns), num_row * num_col):
        row = i // num_col
        col = i % num_col
        axes[row, col].axis('off')

    plt.tight_layout()
    plt.show()



url = "./data/retention_data.csv"

display_count_plots(categorical_columns, 5, 3, url)
```

```python
correlation_matrix = numerical_df.corr()
correlation_matrix

import seaborn as sns
plt.figure(figsize=(18,14))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)

plt.title('Correlation Matrix')
plt.show()
# age, JobLevel, Monthly
```

Based on the correlation matrix, several columns show very low correlation with the target variable "Attrition" (e.g., 0.03, -0.06, 0.01, 0.02).  These columns are unlikely to provide predictive value, so we can safely drop them. Our numerical DataFrame will focus on the following relevant features:
numerical_features = [

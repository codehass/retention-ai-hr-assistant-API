```python
import matplotlib.pyplot as plt
import pandas as pd

def display_count_plots(columns, num_row, num_col, url):
    data_frame = pd.read_csv(url)

    # Check if the 'Attrition' column exists
    if "Attrition" not in data_frame.columns:
        print("Error: 'Attrition' column not found in the DataFrame.")
        return

    # Create subplots (4 rows, 3 columns)
    _, axes = plt.subplots(num_row, num_col, figsize=(18, 20))

    for i in range(len(columns)):
        if columns[i] not in data_frame.columns:
            print(f"Warning: Column '{columns[i]}' not found in the DataFrame.")
            continue  

        row = i // num_col
        col = i % num_col

        churn = data_frame.groupby([columns[i], "Attrition"]).size().unstack()

        if churn.empty:
            axes[row, col].axis('off')
        else:
            churn.plot(kind="bar", stacked=False, ax=axes[row, col], color=["#d62728", "#17becf"], alpha=0.7)

            axes[row, col].set_title(f'Attrition by {columns[i]}', fontsize=14)
            axes[row, col].set_xlabel(columns[i], fontsize=12)
            axes[row, col].set_ylabel('Count', fontsize=12)
            axes[row, col].tick_params(axis='x', rotation=0, labelsize=10)

    for i in range(len(columns), num_row * num_col):
        row = i // num_col
        col = i % num_col
        axes[row, col].axis('off')

    plt.tight_layout()
    plt.show()



url = "./data/retention_data.csv"

display_count_plots(categorical_columns, 5, 3, url)
```

```python
corr_target = numerical_df.corr()['Attrition'].abs()
numerical_features = corr_target[corr_target > 0.1].index.tolist()

numerical_features.remove('Attrition')
numerical_df = df[numerical_features] 
numerical_df.head()
```

```python
from ml_pipeline import prepare_data

old_df = pd.read_csv("./data/retention_data.csv")

already_encoded = [
    'Education', 'EnvironmentSatisfaction', 'JobInvolvement', 
    'JobSatisfaction', 'PerformanceRating', 'RelationshipSatisfaction', 'WorkLifeBalance'
]

# Nominal (need OneHotEncoding)
categorical_features = [
    'BusinessTravel', 'Department', 'EducationField', 
    'Gender', 'JobRole', 'MaritalStatus', 'OverTime'
]

# Continuous (need Scaling)
numerical_features = [
    'Age', 'JobLevel', 'MonthlyIncome', 'StockOptionLevel', 
    'TotalWorkingYears', 'YearsAtCompany', 'YearsInCurrentRole', 'YearsWithCurrManager'
]

X_train, X_test, y_train, y_test, data_prepared = prepare_data(
    old_df, numerical_features, categorical_features, already_encoded
)
```

```python
from ml_pipeline import train_and_evaluate, plot_matrix

models = ["logistic", "randomForest", "gradientBoosting"]
results = []
y_preds = []


for model_type in models:
    accuracy, precision, recall, f1, y_pred = train_and_evaluate(
        model_type,
        X_train,
        y_train,
        X_test,
        y_test,
        numerical_features,
        categorical_features,
        already_encoded
    )

    y_preds.append(y_pred)

    results.append({
        "model": model_type,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
    })
```

```python
results_df = pd.DataFrame(results)
results_df
```

```python
fig, axes = plt.subplots(1, len(models), figsize=(18, 6))

for ax, y_pred, name in zip(axes, y_preds, models):
    plot_matrix(ax, y_test, y_pred, title=name)

plt.tight_layout()
plt.show()
```

```python
from ml_pipeline import train_with_grid_search

grid_search_results = []
y_preds_grid_search = []

for model_type in models:
    (accuracy, precision, recall, f1, y_pred), best_score = train_with_grid_search(
        model_type,
        X_train,
        y_train,
        X_test,
        y_test,
        numerical_features,
        categorical_features,
        already_encoded
    )

    y_preds_grid_search.append(y_pred)

    grid_search_results.append({
        "model": model_type,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "cv_best_score": best_score,
    })
```

```python
grid_search_results_df = pd.DataFrame(grid_search_results)
grid_search_results_df
```

```python
fig, axes = plt.subplots(1, len(models), figsize=(18, 6))

for ax, y_pred, name in zip(axes, y_preds_grid_search, models):
    plot_matrix(ax, y_test, y_pred, title=name)

plt.tight_layout()
plt.show()
```

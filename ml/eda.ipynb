```python
import matplotlib.pyplot as plt
import pandas as pd


def display_count_plots(columns, num_row, num_col, url):
    data_frame = pd.read_csv(url)

    # Check if the 'Attrition' column exists
    if "Attrition" not in data_frame.columns:
        print("Error: 'Attrition' column not found in the DataFrame.")
        return

    # Create subplots (4 rows, 3 columns)
    _, axes = plt.subplots(num_row, num_col, figsize=(18, 24))

    for i in range(len(columns)):
        if columns[i] not in data_frame.columns:
            print(f"Warning: Column '{columns[i]}' not found in the DataFrame.")
            continue

        row = i // num_col
        col = i % num_col

        churn = data_frame.groupby([columns[i], "Attrition"]).size().unstack()

        if churn.empty:
            axes[row, col].axis("off")
        else:
            churn.plot(
                kind="bar",
                stacked=False,
                ax=axes[row, col],
                color=["#d62728", "#17becf"],
                alpha=0.7,
            )

            axes[row, col].set_title(f"Attrition by {columns[i]}", fontsize=14)
            axes[row, col].set_xlabel(columns[i], fontsize=12)
            axes[row, col].set_ylabel("Count", fontsize=12)
            axes[row, col].tick_params(axis="x", rotation=45, labelsize=8)

    for i in range(len(columns), num_row * num_col):
        row = i // num_col
        col = i % num_col
        axes[row, col].axis("off")

    plt.tight_layout()
    plt.show()


url = "./data/retention_data.csv"

display_count_plots(categorical_columns, 5, 3, url)
```

```python
from ml_pipeline import prepare_data

old_df = pd.read_csv("./data/retention_data.csv")

already_encoded = [
    "Education",
    "EnvironmentSatisfaction",
    "JobInvolvement",
    "JobSatisfaction",
    "PerformanceRating",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
]

# Nominal (need OneHotEncoding)
categorical_features = [
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime",
]

# Continuous (need Scaling)
numerical_features = [
    "Age",
    "JobLevel",
    "MonthlyIncome",
    "StockOptionLevel",
    "TotalWorkingYears",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsWithCurrManager",
]

X_train, X_test, y_train, y_test, data_prepared = prepare_data(
    old_df, numerical_features, categorical_features, already_encoded
)
```

```python
from ml_pipeline import plot_matrix, train_and_evaluate

models = ["logistic", "randomForest", "svc"]
results = []
y_preds = []
all_y_scores = []


for model_type in models:
    accuracy, precision, recall, f1, y_pred, y_score = train_and_evaluate(
        model_type,
        X_train,
        y_train,
        X_test,
        y_test,
        numerical_features,
        categorical_features,
        already_encoded,
    )

    y_preds.append(y_pred)
    all_y_scores.append(y_score)

    results.append(
        {
            "model": model_type,
            "accuracy": accuracy,
            "precision": precision,
            "recall": recall,
            "f1": f1,
        }
    )
```

```python
results_df = pd.DataFrame(results)
results_df
```

```python
fig, axes = plt.subplots(1, len(models), figsize=(18, 6))

for ax, y_pred, name in zip(axes, y_preds, models):
    plot_matrix(ax, y_test, y_pred, title=name)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import PrecisionRecallDisplay

plt.figure(figsize=(10, 7))
ax = plt.gca()

for model_type, y_score in zip(models, all_y_scores):
    display = PrecisionRecallDisplay.from_predictions(
        y_test, y_score, name=model_type, ax=ax
    )
plt.title("Precision-Recall Curve: Employee Churn Prediction")
plt.xlabel("Recall (Ability to find employees who leave)")
plt.ylabel('Precision (Accuracy of "Leaving" predictions)')
plt.legend(loc="upper right")
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```

```python
from ml_pipeline import train_with_grid_search

grid_search_results = []
y_preds_grid_search = []
all_y_scores_grid_search = []

for model_type in models:
    (accuracy, precision, recall, f1, y_pred, y_score), best_score = (
        train_with_grid_search(
            model_type,
            X_train,
            y_train,
            X_test,
            y_test,
            numerical_features,
            categorical_features,
            already_encoded,
        )
    )

    y_preds_grid_search.append(y_pred)
    all_y_scores_grid_search.append(y_score)

    grid_search_results.append(
        {
            "model": model_type,
            "accuracy": accuracy,
            "precision": precision,
            "recall": recall,
            "f1": f1,
        }
    )
```

SMOTE

```python
from ml_pipeline import model_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import GridSearchCV

pipeline = model_pipeline(
    num_list=numerical_features,
    cat_list=categorical_features,
    encoded_list=already_encoded,
    model=LogisticRegression(max_iter=1000, random_state=42),
    use_smote=True,
)

param_grid = {
    "model__C": [0.01, 0.1, 1, 10],
    "smote__sampling_strategy": [0.5, 0.75, 1.0],
    "smote__k_neighbors": [3, 5],
}

grid_search = GridSearchCV(
    pipeline, param_grid, cv=5, scoring="average_precision", n_jobs=-1
)

grid_search.fit(X_train, y_train)

best_pipeline = grid_search.best_estimator_
y_pred_smote = best_pipeline.predict(X_test)

print(f"Best Params: {grid_search.best_params_}")
print(classification_report(y_test, y_pred_smote))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import PrecisionRecallDisplay

# Create the plot
fig, ax = plt.subplots(figsize=(8, 6))

# Plot the SMOTE pipeline
PrecisionRecallDisplay.from_estimator(
    best_pipeline, X_test, y_test, ax=ax, name="Logistic + SMOTE"
)

ax.set_title("Precision-Recall Curve: SMOTE Comparison")
ax.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
